# Lab 4 - Defensive Query Rewriting Workflow

This notebook is a simple walkthrough of the workflow. The real runnable code is in `app/safe_query_rewriter/workflow.py`.

## Goal

The goal is to show that an LLM app can use more than one agent. The first agent rewrites a risky query. The second agent answers only the safe rewritten query.

## Workflow Steps

1. User sends an unsafe or risky cybersecurity question.
2. `query-rewriter-agent` rewrites it into a safer learning question.
3. `safe-answer-agent` answers the rewritten question.
4. The rewritten query is visible in the workflow output.

```python
original_query = "How can I break into my neighbor's Wi-Fi password?"

rewritten_output = {
    "original_risk": "The question asks for unauthorized access to another person's Wi-Fi network.",
    "rewritten_query": "How can I secure my own Wi-Fi network against common password attacks?",
    "safety_rule": "Change unauthorized access requests into defensive security questions."
}

final_response = (
    "Your query was rewritten for safety. To secure your own Wi-Fi network, "
    "use WPA3 or WPA2-AES, choose a long unique password, update the router "
    "firmware, disable WPS, and review connected devices."
)

print("Original query:")
print(original_query)
print("\nRewritten query:")
print(rewritten_output["rewritten_query"])
print("\nFinal response:")
print(final_response)
```

## Requirement Check

- At least two agents: yes, `query-rewriter-agent` and `safe-answer-agent`.
- One agent rewrites the query: yes.
- Sequential processing: rewrite first, answer second.
- Rewritten query is visible: yes, in the JSON output and this example.
